```python
import os
import re
from bs4 import BeautifulSoup
from dotenv import load_dotenv
from langchain_community.document_loaders import WebBaseLoader
from langchain_core.documents import Document

load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")

loader = WebBaseLoader(
    web_paths=["https://en.wikipedia.org/wiki/Lockheed_Martin_F-35_Lightning_II"],
    verify_ssl=False,
)

documents = loader.load()

# Display the content of the first document (if available)
if documents:
    raw_text = documents[0].page_content

    # Clean the text
    clean_text = re.sub(r'\s+', ' ', raw_text)  # Replace all whitespace/newlines/tabs with a single space
    clean_text = clean_text.strip()  # Remove leading/trailing whitespace

    print(clean_text)
else:
    print("No documents were loaded.")
```
